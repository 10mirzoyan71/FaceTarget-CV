```python
import cv2

# load face detector and camera
detector = cv2.CascadeClassifier(cv2.data.haarcascades + 'haarcascade_frontalface_default.xml')
cam = cv2.VideoCapture(0, cv2.CAP_DSHOW)

print("Prass Q to quit")

while cam.isOpened():
    ret, img = cam.read()
    if not ret:
        break  # frame read failed

    img = cv2.flip(img, 1)  # mirror frame
    gray_frame = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    
    faces = detector.detectMultiScale(gray_frame, 1.1, 5, minSize=(30, 30))
    h, w, _ = img.shape

    for (x, y, sw, sh) in faces:
        cv2.rectangle(img, (x, y), (x + sw, y + sh), (0, 255, 0), 2)  # face box

        # center coords for target (forehead)
        cx = int(x + sw / 2)
        cy = int(y + sh * 0.2)

        cv2.circle(img, (cx, cy), 3, (0, 255, 255), -1)

        # target crosshair setup
        rad = int(sw / 10)
        cv2.circle(img, (cx, cy), rad, (0, 0, 255), 2)
        
        l_len = int(rad / 2)
        cv2.line(img, (cx - rad - l_len, cy), (cx + rad + l_len, cy), (0, 0, 255), 2)
        cv2.line(img, (cx, cy - rad - l_len), (cx, cy + rad + l_len), (0, 0, 255), 2)

        cv2.putText(img, "Target: Forehead", (w - 320, 40), cv2.FONT_HERSHEY_SIMPLEX, 0.7, (0, 255, 255), 2)

    cv2.putText(img, "camera: index=0 | Press 'q' to quit", (10, h - 20), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (255, 255, 255), 1)

    cv2.imshow('Forehead Tracking System (Micro)', img)

    # exit condition
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cam.release()
cv2.destroyAllWindows()
```
